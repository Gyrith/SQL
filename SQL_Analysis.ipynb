# SQL Practice

SQL practice using SQLite and Pandas, covering querying, filtering, aggregation, joins, subqueries, CTEs, and window functions.

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('sales.db')
```

```python
# create dataset
data = {
    "customer": ["Alice", "Bob", "Charlie", "Diana", "Eve",
                 "Frank", "Grace", "Henry", "Ivy", "Jack"],
    "product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone",
                "Monitor", "Tablet", "Laptop", "Phone", "Monitor"],
    "category": ["Electronics"] * 10,
    "quantity": [1, 2, 1, 2, 3, 1, 2, 1, 2, 1],
    "price": [800, 500, 300, 800, 500, 250, 300, 800, 500, 250],
    "region": ["East", "West", "North", "East", "South",
               "West", "North", "South", "East", "West"]
}

df = pd.DataFrame(data)

df
```

```python
df.to_sql('sales',conn,if_exists='replace',index=False)
```

## SELECT

`SELECT` retrieves data from a table.

```python
query = """
       SELECT *
       FROM sales ;
       """
pd.read_sql_query(query,conn)
```

## Selecting Columns

Specific columns can be selected instead of retrieving the entire table.

```python
query = """
       Select customer,product,price
       from sales;
       """
pd.read_sql_query(query,conn)
```

## DISTINCT

`DISTINCT` returns unique values and removes duplicates.

```python
query = """
     SELECT DISTINCT product
     FROM sales;
     """
pd.read_sql_query(query,conn)
```

## WHERE

`WHERE` filters rows based on a condition.

```python
query = """
    SELECT *
    FROM sales
    WHERE price > 500 ;
    """
pd.read_sql_query(query,conn)
```

## AND / OR

`AND` requires all conditions to be true.

`OR` requires at least one condition to be true.

```python
query = """
      SELECT *
      FROM sales
      WHERE price > 300
      AND region ='East' ;
      """
pd.read_sql_query(query,conn)
```

```python
# OR
query = """
SELECT *
FROM sales
WHERE region = 'East'
OR region = 'West';
"""

pd.read_sql_query(query, conn)
```

## IN

`IN` checks whether a value matches any value in a specified list.

```python
query = """
       SELECT *
       FROM sales
       WHERE product IN ('Laptop','Phone') ;
       """
pd.read_sql_query(query,conn)
```

## BETWEEN

`BETWEEN` filters values within a specified range.

```python
query = """
      SELECT *
      FROM sales
      WHERE price BETWEEN 300 AND 800;
      """

pd.read_sql_query(query, conn)
```

## LIKE

`LIKE` searches for a specified pattern in text.

`%` represents zero or more characters.

```python
query = """
      SELECT *
      FROM sales
      WHERE customer LIKE 'A%';
      """

pd.read_sql_query(query, conn)
```

```python
query = """
      SELECT *
      FROM sales
      WHERE customer LIKE '%A';
      """

pd.read_sql_query(query, conn)
```

## ORDER BY

`ORDER BY` sorts query results.

`ASC` sorts in ascending order.

`DESC` sorts in descending order.

```python
query = """
     SELECT *
     FROM sales
     ORDER BY price DESC;
     """

pd.read_sql_query(query, conn)
```

## LIMIT

`LIMIT` restricts the number of rows returned.

```python
query = """
     SELECT *
     FROM sales
     ORDER BY price DESC
     LIMIT 5;
     """

pd.read_sql_query(query, conn)
```

## Aggregate Functions

Aggregate functions perform calculations on multiple rows.

Common aggregate functions include:

- `COUNT()`
- `SUM()`
- `AVG()`
- `MIN()`
- `MAX()`

```python
query = """
    SELECT
    COUNT(*) AS total_sales,
    SUM(quantity) AS total_quantity,
    AVG(price) AS average_price,
    MIN(price) AS minimum_price,
    MAX(price) AS maximum_price
    FROM sales;
    """

pd.read_sql_query(query, conn)
```

## GROUP BY

`GROUP BY` groups rows that have the same value in one or more columns.

It is commonly used with aggregate functions.

```python
query = """
SELECT
    product,
    SUM(quantity) AS total_quantity
FROM sales
GROUP BY product;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    region,
    product,
    SUM(quantity) AS total_quantity
FROM sales
GROUP BY region, product;
"""

pd.read_sql_query(query, conn)
```

## HAVING

`HAVING` filters grouped results.

`WHERE` filters rows before grouping, while `HAVING` filters groups after aggregation.

```python
query = """
SELECT
    product,
    SUM(quantity) AS total_quantity
FROM sales
GROUP BY product
HAVING SUM(quantity) > 2;
"""

pd.read_sql_query(query, conn)
```

## CASE

`CASE` allows conditional logic inside a SQL query.

It works similarly to an `if / elif / else` structure.

```python
query = """
SELECT
    customer,
    product,
    price,
    CASE
        WHEN price >= 700 THEN 'High'
        WHEN price >= 400 THEN 'Medium'
        ELSE 'Low'
    END AS price_category
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## Calculated Columns

SQL can create new values from existing columns.

```python
query = """
SELECT
    customer,
    product,
    quantity,
    price,
    quantity * price AS total_sales
FROM sales;
"""

pd.read_sql_query(query, conn)
```

```python
customers = pd.DataFrame({
    "customer": ["Alice", "Bob", "Charlie", "Diana", "Eve",
                 "Frank", "Grace", "Henry", "Ivy", "Jack"],
    "age": [25, 32, 28, 41, 35, 29, 38, 45, 27, 33],
    "membership": ["Gold", "Silver", "Gold", "Gold", "Silver",
                   "Bronze", "Gold", "Silver", "Bronze", "Gold"]
})

customers.to_sql("customers", conn, if_exists="replace", index=False)

customers
```

## INNER JOIN

An `INNER JOIN` returns rows where there is a matching value in both tables.

```python
query = """
SELECT
    sales.customer,
    sales.product,
    sales.price,
    customers.membership
FROM sales
INNER JOIN customers
ON sales.customer = customers.customer;
"""

pd.read_sql_query(query, conn)
```

## LEFT JOIN

A `LEFT JOIN` returns all rows from the left table and matching rows from the right table.

If there is no match, the right-side columns contain `NULL`.

```python
query = """
SELECT
    sales.customer,
    sales.product,
    customers.membership
FROM sales
LEFT JOIN customers
ON sales.customer = customers.customer;
"""

pd.read_sql_query(query, conn)
```

## RIGHT JOIN

A `RIGHT JOIN` returns all rows from the right table and matching rows from the left table.

```python
query = """
SELECT
    customers.customer,
    customers.membership,
    sales.product
FROM sales
RIGHT JOIN customers
ON sales.customer = customers.customer;
"""

pd.read_sql_query(query, conn)
```

## FULL OUTER JOIN

A `FULL OUTER JOIN` returns all rows from both tables, matching rows where possible.

```python
query = """
SELECT
    customers.customer,
    customers.membership,
    sales.product
FROM sales
FULL OUTER JOIN customers
ON sales.customer = customers.customer;
"""

pd.read_sql_query(query, conn)
```

## CROSS JOIN

A `CROSS JOIN` returns every possible combination of rows from two tables.

```python
query = """
SELECT
    sales.product,
    customers.membership
FROM sales
CROSS JOIN customers;
"""

pd.read_sql_query(query, conn)
```

```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4],
    "employee_name": ["Alice", "Bob", "Charlie", "Diana"],
    "manager_id": [None, 1, 1, 2]
})

employees.to_sql("employees", conn, if_exists="replace", index=False)
```

## SELF JOIN

A self join joins a table with itself.

It can be useful for relationships such as employees and their managers.

```python
query = """
SELECT
    e.employee_name AS employee,
    m.employee_name AS manager
FROM employees e
LEFT JOIN employees m
ON e.manager_id = m.employee_id;
"""

pd.read_sql_query(query, conn)
```

## Subquery

A subquery is a query nested inside another query.

It can be used when the result of one query is needed by another query.

```python
query = """
SELECT *
FROM sales
WHERE price > (
    SELECT AVG(price)
    FROM sales
);
"""

pd.read_sql_query(query, conn)
```

## Common Table Expression

A CTE temporarily stores the result of a query using `WITH`.

CTEs can make complex queries easier to read and organize.

```python
query = """
WITH product_sales AS (
    SELECT
        product,
        SUM(quantity * price) AS total_sales
    FROM sales
    GROUP BY product
)

SELECT *
FROM product_sales
WHERE total_sales > 1000;
"""

pd.read_sql_query(query, conn)
```

## Window Functions

Window functions perform calculations across related rows without collapsing them into groups.

They are useful for rankings, running totals, comparisons, and other analytical calculations.

```python
query = """
SELECT
    customer,
    product,
    price,
    RANK() OVER (ORDER BY price DESC) AS price_rank
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## PARTITION BY

`PARTITION BY` divides rows into groups before a window function is applied.

This allows calculations such as ranking within each category.

```python
query = """
SELECT
    region,
    customer,
    price,
    RANK() OVER (
        PARTITION BY region
        ORDER BY price DESC
    ) AS regional_rank
FROM sales;
"""

pd.read_sql_query(query, conn)
```

# Mini Analysis

Using the sales dataset, we can identify total revenue by product and determine which products generate the most sales.

```python
query = """
SELECT
    product,
    SUM(quantity) AS units_sold,
    SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
ORDER BY revenue DESC;
"""

pd.read_sql_query(query, conn)
```

## NULL Handling

`NULL` represents a missing or unknown value.

Common SQL tools for handling NULL values include:
- `IS NULL`
- `IS NOT NULL`
- `COALESCE()`
- `NULLIF()`

```python
query = """
SELECT *
FROM customers
WHERE membership IS NULL;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    customer,
    COALESCE(membership, 'Unknown') AS membership
FROM customers;
"""

pd.read_sql_query(query, conn)
```

## UNION

`UNION` combines the results of two queries and removes duplicate rows.

```python
query = """
SELECT customer
FROM sales
WHERE region = 'East'

UNION

SELECT customer
FROM sales
WHERE region = 'West';
"""

pd.read_sql_query(query, conn)
```

## UNION ALL

`UNION ALL` combines query results while keeping duplicates.

It is generally faster than `UNION` because duplicate removal is not performed.

```python
query = """
SELECT customer
FROM sales
WHERE region = 'East'

UNION ALL

SELECT customer
FROM sales
WHERE region = 'West';
"""

pd.read_sql_query(query, conn)
```

## String Functions

SQL provides functions for manipulating text data.

Examples include:
- `UPPER()`
- `LOWER()`
- `LENGTH()`
- `SUBSTR()`

```python
query = """
SELECT
    customer,
    UPPER(customer) AS uppercase_name,
    LOWER(customer) AS lowercase_name,
    LENGTH(customer) AS name_length
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## Date Functions

SQL can extract and manipulate information from date values.

```python
query = """
SELECT
    customer,
    sale_date,
    strftime('%Y', sale_date) AS year,
    strftime('%m', sale_date) AS month
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## CAST

`CAST()` converts a value from one data type to another.

```python
query = """
SELECT
    customer,
    CAST(price AS REAL) AS price
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## ROW_NUMBER()

`ROW_NUMBER()` assigns a unique sequential number to each row within a window.

```python
query = """
SELECT
    customer,
    product,
    price,
    ROW_NUMBER() OVER (
        ORDER BY price DESC
    ) AS row_number
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## LAG()

`LAG()` accesses a value from a previous row.

It is useful for comparing the current row with a previous value.

```python
query = """
SELECT
    sale_date,
    price,
    LAG(price) OVER (
        ORDER BY sale_date
    ) AS previous_price
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## LEAD()

`LEAD()` accesses a value from a following row.

It is useful for comparing the current row with a future value.

```python
query = """
SELECT
    sale_date,
    price,
    LEAD(price) OVER (
        ORDER BY sale_date
    ) AS next_price
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## Running Total

Window functions can be used to calculate cumulative values across ordered rows.

```python
query = """
SELECT
    sale_date,
    quantity,
    SUM(quantity) OVER (
        ORDER BY sale_date
    ) AS running_quantity
FROM sales;
"""

pd.read_sql_query(query, conn)
```

## Conditional Aggregation

Conditional aggregation combines aggregate functions with `CASE` to calculate values based on conditions.

```python
query = """
SELECT
    region,
    SUM(
        CASE
            WHEN price >= 500 THEN quantity
            ELSE 0
        END
    ) AS high_value_units
FROM sales
GROUP BY region;
"""

pd.read_sql_query(query, conn)
```
